```python
import os
import h5py

base_dir = "/work/dlclarge2/zhangj-zhangj-CFM/data/training2"
bad_files = []

for root, _, files in os.walk(base_dir):
    for f in files:
        if f.endswith(".h5"):
            path = os.path.join(root, f)
            try:
                with h5py.File(path, "r") as h5f:
                    _ = list(h5f.keys())  # 尝试读取文件结构
            except Exception as e:
                print(f"Corrupted file: {path}")
                bad_files.append(path)

print(f"\nTotal bad files: {len(bad_files)}")
```
